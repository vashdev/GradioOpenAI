# Creating Designs by Leveraging OpenAI and Gradio UI

This notebook demonstrates how to build a web-based platform that generates bespoke images and designs for banners and posters using OpenAI's DALL-E API and Gradio UI.

```python
import os
import gradio as gr
import requests
from PIL import Image
from io import BytesIO
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()
```

```python
# Initialize the OpenAI client — reads OPENAI_API_KEY from .env
client = OpenAI(api_key=os.environ["OPENAI_API_KEY"])
```

```python
def generate_image(prompt):
    """
    Generate an image from a text prompt using OpenAI's DALL-E model.

    Args:
        prompt (str): Text description of the image to generate.

    Returns:
        PIL.Image: The generated image as a PIL Image object.
    """
    response = client.images.generate(
        model="dall-e-3",
        prompt=prompt,
        size="1024x1024",
        quality="standard",
        n=1
    )

    image_url = response.data[0].url
    image_response = requests.get(image_url)
    image = Image.open(BytesIO(image_response.content))

    return image
```

```python
# Set up the Gradio interface
interface = gr.Interface(
    fn=generate_image,
    inputs=gr.Textbox(
        lines=3,
        placeholder="Enter your design prompt here... e.g., 'A dramatic Netflix banner for a sci-fi thriller set in outer space'",
        label="Design Prompt"
    ),
    outputs=gr.Image(
        type="pil",
        label="Generated Design"
    ),
    title="Netflix Campaign Design Generator",
    description=(
        "Generate bespoke images and designs for Netflix banners and posters. "
        "Enter a descriptive text prompt and let OpenAI's DALL-E transform your ideas into "
        "visually compelling promotional designs."
    ),
    examples=[
        ["A dramatic Netflix banner featuring a lone warrior standing on a dystopian cityscape at sunset, cinematic lighting, 4K quality"],
        ["A Netflix movie poster for a romantic comedy, warm pastel colors, couple silhouette against the Eiffel Tower at night"],
        ["A dark and atmospheric Netflix thriller poster with shadowy figures in a misty forest, deep blues and greens"]
    ],
    theme=gr.themes.Soft()
)
```

```python
# Launch the Gradio interface
interface.launch(share=True)
```
